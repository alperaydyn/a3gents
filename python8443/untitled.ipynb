# PostgreSQL Database Connection
Connect to the `postgres` container running on the shared Docker network.

```python
import os
from dotenv import load_dotenv

load_dotenv()  # loads .env from the working directory if present
```

```python
import psycopg2
from psycopg2.extras import RealDictCursor


def connect_to_db(
    host: str = os.getenv("POSTGRES_HOST", "postgres"),
    port: int = int(os.getenv("POSTGRES_PORT", 5432)),
    dbname: str = os.getenv("POSTGRES_DB", "postgres"),
    user: str = os.getenv("POSTGRES_USER", "postgres"),
    password: str = os.getenv("POSTGRES_PASSWORD", ""),
):
    """Connect to the PostgreSQL container and return a connection object.

    Parameters
    ----------
    host     : Docker container name (or IP). Defaults to 'postgres' via env.
    port     : Port inside the shared Docker network (always 5432 internally).
    dbname   : Database name from POSTGRES_DB env var.
    user     : Username from POSTGRES_USER env var.
    password : Password from POSTGRES_PASSWORD env var.

    Returns
    -------
    psycopg2.connection
        An open database connection. Call .close() when done, or use
        the connection as a context manager (``with connect_to_db() as conn:``).
    """
    conn = psycopg2.connect(
        host=host,
        port=port,
        dbname=dbname,
        user=user,
        password=password,
    )
    print(f"✅ Connected to '{dbname}' on {host}:{port} as '{user}'")
    return conn


# ── quick test ────────────────────────────────────────────────────────────────
conn = connect_to_db()
with conn.cursor(cursor_factory=RealDictCursor) as cur:
    cur.execute("SELECT version();")
    row = cur.fetchone()
    print("PostgreSQL version:", row["version"])
conn.close()
```

```python
# ── Example: list tables in the public schema ─────────────────────────────────
conn = connect_to_db()
with conn.cursor(cursor_factory=RealDictCursor) as cur:
    cur.execute("""
        SELECT table_name
        FROM information_schema.tables
        WHERE table_schema = 'public'
        ORDER BY table_name;
    """)
    tables = cur.fetchall()

print("Tables:", [t["table_name"] for t in tables])
conn.close()
```
